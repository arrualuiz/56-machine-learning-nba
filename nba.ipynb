# Atividade Somativa 1 - Técnicas de Machine Learning

# Introdução

O objetivo deste trabalho é utilizar técnicas de Machine Learning para prever a variável WinShares presente no dataset NBA Players Stats.

A variável WinShares representa uma estimativa da contribuição de um jogador para as vitórias de sua equipe ao longo de uma temporada.

Como a variável alvo é numérica e contínua, este problema é classificado como um problema de regressão supervisionada.

```python
#importando planilha csv para jupyter notebook
#biblioteca pandas ótima para lidar com planilhas
#salva a planilha em df

import pandas as pd

df = pd.read_csv('nba_stats.csv')

print(df.shape)

df.head()
```

# Análise Inicial dos Dados

Nesta etapa foi realizado o carregamento do dataset e uma análise preliminar da estrutura dos dados, incluindo quantidade de registros, tipos de variáveis e identificação de valores ausentes.

```python
#informações gerais da planilha
df.info()
```

```python
df.isnull().sum()
#valores ausentes
```

# Preparação dos Dados

Foi realizada uma análise dos valores ausentes presentes no dataset. Como a quantidade de registros com valores faltantes representa uma pequena parcela da base de dados, optou-se pela remoção dessas observações para simplificar o processo de preparação dos dados.

Além disso, a variável categórica "Posicao" foi convertida para formato numérico utilizando a técnica One-Hot Encoding, permitindo sua utilização pelos algoritmos de Machine Learning.

```python
#removendo registros com valores ausentes

df = df.dropna()

print(df.shape)
```

```python
#ajustando categórica "Posicao" para formato numérico

df = pd.get_dummies(df, columns=['Posicao'])

print(df.shape)
df.head()
```

# Definição das Variáveis

A variável WinShares foi definida como variável alvo, ou seja, o valor que o modelo buscará prever.

As demais variáveis do dataset foram utilizadas como atributos preditores.

```python
#separação entre variáveis preditoras (X) e variável alvo (y)

X = df.drop('WinShares', axis=1)
y = df['WinShares']

print(X.shape)
print(y.shape)
```

# Seleção de Atributos

O enunciado solicita a aplicação obrigatória de uma técnica de seleção ou extração de atributos.

Neste trabalho foi utilizada a técnica SelectKBest com a função estatística f_regression. Essa técnica seleciona os atributos que apresentam maior relação estatística com a variável alvo WinShares.

```python
#importação da técnica de seleção de atributos
#utilizando a função estatística f_regression com 10 atributos

from sklearn.feature_selection import SelectKBest, f_regression

selector = SelectKBest(score_func=f_regression, k=10)

X_selected = selector.fit_transform(X, y)

print(X_selected.shape)
```

```python
#exibição dos atributos escolhidos pela técnica SelectKBest
selected_columns = X.columns[selector.get_support()]

selected_columns
```

# Divisão entre Treino e Teste

Após a preparação dos dados, a base foi dividida em 75% para treinamento e 25% para teste, conforme solicitado no enunciado da atividade.

A base de treinamento foi utilizada para construir o modelo de Machine Learning e a base de teste foi utilizada para avaliar sua capacidade de previsão em dados não utilizados durante o treinamento.

```python
#divisão da base de dados em treino e teste, 25% teste 75% treinamento
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y,
    test_size=0.25,
    random_state=42
)

print("Treino:", X_train.shape)
print("Teste:", X_test.shape)
```

# Treinamento do Modelo

Após a preparação dos dados e a seleção dos atributos mais relevantes, foi utilizado o algoritmo Random Forest Regressor para realizar o treinamento do modelo.

Esse algoritmo foi escolhido por sua capacidade de lidar com relações não lineares entre variáveis e por apresentar bons resultados em problemas de regressão.

```python
#importação do algoritmo Random Forest Regressor
from sklearn.ensemble import RandomForestRegressor

#criação do modelo
modelo = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
#treinamento do modelo utilizando a base de treino
modelo.fit(X_train, y_train)
```

# Predição da Base de Teste

Após o treinamento, o modelo foi utilizado para realizar previsões sobre a base de teste, composta por 25% dos registros do dataset.

```python
#realização das previsões utilizando a base de teste
predicoes = modelo.predict(X_test)

predicoes[:10]
```

# Avaliação do Modelo

Para avaliar o desempenho do modelo foram utilizadas métricas de regressão.

As métricas escolhidas foram:

- MAE: erro absoluto médio
- RMSE: raiz do erro quadrático médio
- R²: coeficiente de determinação

```python
#importação das métricas de avaliação
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

#calculo das metricas do modelo
mae = mean_absolute_error(y_test, predicoes)

rmse = mean_squared_error(y_test, predicoes) ** 0.5

r2 = r2_score(y_test, predicoes)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```

# Comparação entre Valores Reais e Previstos

A tabela abaixo apresenta uma comparação entre os valores reais de WinShares e os valores previstos pelo modelo para os primeiros registros da base de teste.

```python
#comparação entre valores reais e valores previstos
resultado = pd.DataFrame({
    'Valor Real': y_test.values,
    'Valor Previsto': predicoes
})

resultado.head(20)
```

# Análise dos Resultados

Ao comparar os valores reais com os valores previstos pelo modelo, observa-se que as previsões ficaram muito próximas dos valores originais do dataset.

Por exemplo, para um jogador com WinShares real igual a 7,2, o modelo previu 7,206. Da mesma forma, para um valor real de 8,8, a previsão foi de 8,641. Essas diferenças são pequenas e indicam que o modelo conseguiu identificar adequadamente os padrões presentes nos dados.

Os resultados observados na amostra apresentada corroboram as métricas calculadas anteriormente, especialmente o alto valor de R² (0,9987) e os baixos valores de erro (MAE e RMSE), indicando que o modelo apresentou alta capacidade preditiva para os dados utilizados neste estudo.
